## Exploratory Data Analysis (EDA) - Accepted Loans Dataset

```python
overview = pd.DataFrame({
    'dtype': df.dtypes,
    'non_null_count': df.count(),
    'null_pct': (df.isnull().mean() * 100).round(1),
    'unique_values': df.nunique(),
    'sample_value': df.iloc[0]
})
print(overview)
```
