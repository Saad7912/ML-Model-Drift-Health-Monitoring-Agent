# Corelation between Features

```python
# Select only numerical columns
numerical_df = df.select_dtypes(include=['number'])

# Compute correlation matrix
corr_matrix = numerical_df.corr()

# Display correlation matrix
print(corr_matrix)
```

```python
import numpy as np

# Set correlation threshold
threshold = 0.85

# Get upper triangle of the correlation matrix
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Find highly correlated feature pairs
high_corr_pairs = []

for column in upper_triangle.columns:
    for row in upper_triangle.index:
        corr_value = upper_triangle.loc[row, column]

        if pd.notna(corr_value) and abs(corr_value) > threshold:
            high_corr_pairs.append({
                "Feature 1": row,
                "Feature 2": column,
                "Correlation": round(corr_value, 3)
            })

high_corr_df = pd.DataFrame(high_corr_pairs)

print(high_corr_df.sort_values(by="Correlation", ascending=False))
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

chi_square_results = []

for feature in categorical_features:
    
    # Create contingency table
    contingency_table = pd.crosstab(df[feature], df["target"])
    
    # Chi-square test
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    chi_square_results.append({
        "Feature": feature,
        "Chi-Square": chi2,
        "P-Value": p_value
    })

# Convert results into DataFrame
chi_square_df = pd.DataFrame(chi_square_results)

# Sort by p-value
chi_square_df = chi_square_df.sort_values("P-Value")

print(chi_square_df)
```

# **Rough Interpretation of Cramers'V**
|  Cramér's V | Relationship |
| ----------: | ------------ |
|    0 – 0.10 | Very weak    |
| 0.10 – 0.20 | Weak         |
| 0.20 – 0.30 | Moderate     |
|       0.30+ | Stronger     |


```python
import numpy as np
from scipy.stats import chi2_contingency

cramers_results = []

for feature in categorical_features:
    
    contingency_table = pd.crosstab(df[feature], df["target"])
    
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    n = contingency_table.sum().sum()
    r, k = contingency_table.shape
    
    cramers_v = np.sqrt(
        (chi2 / n) / min(k - 1, r - 1)
    )
    
    cramers_results.append({
        "Feature": feature,
        "Chi-Square": chi2,
        "P-Value": p_value,
        "Cramer's V": cramers_v
    })

cramers_df = pd.DataFrame(cramers_results)

cramers_df = cramers_df.sort_values(
    "Cramer's V",
    ascending=False
)

print(cramers_df)
```

# **Mutual Information Testing**

```python
X_train = train_2014[candidate_features].copy()
y_train = train_2014["target"].copy()

print(X_train.shape)
print(y_train.shape)
```

Seprate numerical and catagorical features in it 

```python
categorical_features_mi = [
    "application_type",
    "home_ownership",
    "initial_list_status",
    "purpose",
    "verification_status"
]

numerical_features_mi = [
    col for col in candidate_features
    if col not in categorical_features_mi
]
```

Handle missing value

Encode catagorical values

```python
X_mi = X_train.copy()

for col in numerical_features_mi:
    median = X_mi[col].median()

    if pd.isna(median):
        X_mi[col] = 0
    else:
        X_mi[col] = X_mi[col].fillna(median)

for col in categorical_features_mi:
    X_mi[col] = X_mi[col].fillna("Missing")
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.feature_selection import mutual_info_classif
```

```python
# -----------------------------
# 4. Encode categorical features
# -----------------------------
encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_mi[categorical_features_mi] = encoder.fit_transform(
    X_mi[categorical_features_mi]
)
```

```python
print("Remaining NaN:", X_mi.isna().sum().sum())
```

```python
# -----------------------------
# 6. Mutual Information
# -----------------------------
categorical_indices = [
    X_mi.columns.get_loc(col)
    for col in categorical_features_mi
]

discrete_mask = [
    i in categorical_indices
    for i in range(X_mi.shape[1])
]

mi_scores = mutual_info_classif(
    X_mi,
    y_train,
    discrete_features=discrete_mask,
    random_state=42
)
```

```python
# -----------------------------
# 7. Results
# -----------------------------
mi_results = pd.DataFrame({
    "Feature": X_mi.columns,
    "MI_Score": mi_scores
}).sort_values(
    "MI_Score",
    ascending=False
)

print(mi_results.to_string(index=False))
```

# **Random Forest Feature Importance**

```python
from sklearn.ensemble import RandomForestClassifier

# Sample training data for feature importance
X_rf = X_mi.sample(
    n=150000,
    random_state=42
)

y_rf = y_train.loc[X_rf.index]

rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf.fit(X_rf, y_rf)
```

```python
rf_results = pd.DataFrame({
    "Feature": X_rf.columns,
    "RF_Importance": rf.feature_importances_
}).sort_values(
    "RF_Importance",
    ascending=False
)

print(rf_results.to_string(index=False))
```

Combine this MI and RF tables

```python
combined_df = pd.merge(
    mi_results,
    rf_results,
    on="Feature",
    how="outer"
)

combined_df = combined_df.sort_values(
    "RF_Importance",
    ascending=False
).reset_index(drop=True)

print(combined_df.to_string(index=False))
```

# **Model Training**

Splitting Data

```python
train_2014 = df[df["issue_d"].dt.year <= 2014].copy()
test_2015 = df[df["issue_d"].dt.year == 2015].copy()

X_train = train_2014[final_candidate_features]
y_train = train_2014["target"]

X_test = test_2015[final_candidate_features]
y_test = test_2015["target"]

print("Training:", X_train.shape)
print("Testing :", X_test.shape)
```

Seprate numerical and catagorical features

```python
categorical_features = [
    col for col in [
        "application_type",
        "home_ownership",
        "initial_list_status",
        "purpose",
        "verification_status"
    ]
    if col in final_candidate_features
]
```

```python
numerical_features = [
    col for col in final_candidate_features
    if col not in categorical_features
]

print("Numerical:", len(numerical_features))
print("Categorical:", len(categorical_features))
```

PreProcessing

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numerical_features),
    ("cat", categorical_transformer, categorical_features)
])
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        n_jobs=-1
    ))
])
```

Train

```python
logistic_model.fit(X_train, y_train)
```

Predict

```python
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

Evaluate

```python
from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("ROC-AUC:", roc_auc_score(y_test, y_prob))

print("PR-AUC:", average_precision_score(y_test, y_prob))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
